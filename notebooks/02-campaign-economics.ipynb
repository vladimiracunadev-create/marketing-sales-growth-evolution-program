# Campaign economics
Dataset: `campaigns.csv`. Work from the business question, then define the metric before interpreting results.


```python
import pandas as pd
df=pd.read_csv('../datasets/campaigns.csv')
df['ctr']=df.clicks/df.impressions
df['cpc']=df.spend_clp/df.clicks
df['cpa']=df.spend_clp/df.conversions.replace(0,pd.NA)
df['roas']=df.revenue_clp/df.spend_clp
df.groupby('channel')[['spend_clp','revenue_clp']].sum()
```
