# A/B testing
Dataset: `experiments.csv`. Work from the business question, then define the metric before interpreting results.


```python
import pandas as pd
df=pd.read_csv('../datasets/experiments.csv')
df['control_rate']=df.control_conversions/df.control_n
df['variant_rate']=df.variant_conversions/df.variant_n
df['uplift']=(df.variant_rate/df.control_rate)-1
df[['experiment_id','area','control_rate','variant_rate','uplift']].head(10)
```
