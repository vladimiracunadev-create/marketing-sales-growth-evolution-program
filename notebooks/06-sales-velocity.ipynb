# Sales velocity
Dataset: `leads.csv`. Work from the business question, then define the metric before interpreting results.


```python
import pandas as pd
df=pd.read_csv('../datasets/leads.csv')
opp=df[df.status.isin(['Opportunity','Won','Lost'])]
print('Opportunity records:',len(opp))
print('Won:',(opp.status=='Won').sum())
print('Indicative win rate:',(opp.status=='Won').mean())
```
