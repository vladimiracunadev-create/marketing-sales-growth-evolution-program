# Cohort thinking
Dataset: `customers.csv`. Work from the business question, then define the metric before interpreting results.


```python
import pandas as pd
df=pd.read_csv('../datasets/customers.csv')
df['tenure_band']=pd.cut(df.tenure_months,[0,6,12,24,36,60])
df.groupby('tenure_band',observed=True).agg(n=('customer_id','count'), churn=('churned','mean'), mrr=('mrr_clp','mean'))
```
